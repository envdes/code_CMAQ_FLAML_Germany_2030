```python
import matplotlib.pyplot as plt
import numpy as np

# Shared style of both panels.
COLOR_CMAQ = "#E8705F"        # Reference model (salmon red).
COLOR_FLAML = "#5B9BD5"       # CMAQ-FLAML (light blue).
EDGE_COLOR = "black"
TITLE_SIZE = 15
LABEL_SIZE = 13
TICK_SIZE = 12
VALUE_SIZE = 13
```

```python
# (a) Error metrics under heavy pollution.
metrics = ["RMSE", "MAE"]
cmaq_values = [65.46, 63.05]          # CMAQ error (ug/m3).
flaml_values = [18.94, 15.16]         # CMAQ-FLAML error (ug/m3).

x = np.arange(len(metrics))
width = 0.35

fig, ax = plt.subplots(figsize=(7.2, 4.8))

bars_cmaq = ax.bar(
    x - width / 2, cmaq_values, width,
    label="CMAQ", color=COLOR_CMAQ, edgecolor=EDGE_COLOR, linewidth=1.0
)
bars_flaml = ax.bar(
    x + width / 2, flaml_values, width,
    label="CMAQ-FLAML", color=COLOR_FLAML, edgecolor=EDGE_COLOR, linewidth=1.0
)

# Value labels above the bars.
for bar, value in list(zip(bars_cmaq, cmaq_values)) + list(zip(bars_flaml, flaml_values)):
    ax.annotate(
        f"{value:.2f}",
        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 4),
        textcoords="offset points",
        ha="center", va="bottom",
        fontsize=VALUE_SIZE, fontweight="bold",
    )

ax.set_title("(a) Error Metrics (Heavy Pollution)", fontsize=TITLE_SIZE, fontweight="bold")
ax.set_xlabel("Metrics", fontsize=LABEL_SIZE)
ax.set_ylabel(r"Error ($\mu$g/m$^3$)", fontsize=LABEL_SIZE)

ax.set_xticks(x)
ax.set_xticklabels(metrics, fontsize=TICK_SIZE)
ax.set_ylim(0, 85)
ax.set_yticks(np.arange(0, 81, 10))
ax.tick_params(axis="y", labelsize=TICK_SIZE)

ax.grid(True, linestyle="--", alpha=0.5)
ax.set_axisbelow(True)
ax.legend(fontsize=TICK_SIZE, loc="upper right", frameon=True)

plt.tight_layout()
plt.show()
```

```python
# (b) Exceedance day ratio.
labels = ["Observation", "CMAQ-FLAML", "CMAQ"]
ratios = [15.95, 14.66, 0.00]         # Percentage of exceedance days.
colors = [COLOR_CMAQ, COLOR_FLAML, COLOR_CMAQ]

fig, ax = plt.subplots(figsize=(5.6, 5.4))

bars = ax.bar(
    np.arange(len(labels)), ratios, 0.5,
    color=colors, edgecolor=EDGE_COLOR, linewidth=1.0
)

# Value labels above the bars.
for bar, value in zip(bars, ratios):
    ax.annotate(
        f"{value:.2f}%",
        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 4),
        textcoords="offset points",
        ha="center", va="bottom",
        fontsize=VALUE_SIZE, fontweight="bold",
    )

ax.set_title("(b) Exceedance Day Ratio", fontsize=TITLE_SIZE, fontweight="bold")
ax.set_ylabel("Ratio", fontsize=LABEL_SIZE)

ax.set_xticks(np.arange(len(labels)))
ax.set_xticklabels(labels, fontsize=TICK_SIZE)
ax.set_ylim(0, 21)
ax.set_yticks(np.arange(0, 20.1, 2.5))
ax.tick_params(axis="y", labelsize=TICK_SIZE)

ax.grid(True, linestyle="--", alpha=0.5)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
